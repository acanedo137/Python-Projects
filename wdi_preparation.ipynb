## Infrastructure Data

```python
# TODO 3: Place your solution here.
with open("C:\\Users\\acane\\OneDrive\\Desktop\\CMU\\Python Projects\\ppp-p8-data-analysis\\wdi_data\\2017infra.json", 'r') as infra:
    infra_data = json.load(infra)

requested_metrics = [
    'Air transport, passengers carried',
    'Fixed broadband subscriptions (per 100 people)',
    'Individuals using the Internet (% of population)', 
    'High-technology exports (% of manufactured exports)',
    'Research and development expenditure (% of GDP)',
    'Technicians in R&D (per million people)'
]

renamed_metrics = {
    'Air transport, passengers carried':'Air travel passengers',
    'Fixed broadband subscriptions (per 100 people)': 'Broadband subscriptions (per 100 people)',
    'Individuals using the Internet (% of population)':'Internet use (% of population)', 
    'High-technology exports (% of manufactured exports)':'High-tech exports (%)',
    'Research and development expenditure (% of GDP)':'R&D expenditure (% of GDP)',
    'Technicians in R&D (per million people)':'Technicians in R&D (per million people)'
}

flattened_data = {}
for country, categories in infra_data.items():
    country_metrics  = {}
    for category, metrics in categories.items():
        country_metrics.update(metrics)
    flattened_data[country] = country_metrics

infrastructure_data = pd.DataFrame.from_dict(flattened_data, orient='index')
infrastructure_data = infrastructure_data[requested_metrics]
infrastructure_data = infrastructure_data.rename(columns=renamed_metrics)

print(infrastructure_data)
```

## Poverty Data

```python
# TODO 4: Place your solution here.
poverty_path = pd.read_xml("C:\\Users\\acane\\OneDrive\\Desktop\\CMU\\Python Projects\\ppp-p8-data-analysis\\wdi_data\\World_Development_Indicators_Poverty_2017.xml", xpath="//country")

renamed_metrics = {
    "Gini index (World Bank estimate)": "Gini index",
    "Income share held by highest 10%": "Income share of highest 10%",
    "Survey mean consumption or income per capita, total population (2011 PPP $ per day)": "Daily consumption per capita (USD)",
}

tree = ET.parse("C:\\Users\\acane\\OneDrive\\Desktop\\CMU\\Python Projects\\ppp-p8-data-analysis\\wdi_data\\World_Development_Indicators_Poverty_2017.xml")
root = tree.getroot()
country_datalist = []
for country_elem in root.findall('country'):
    country_name = country_elem.get('name')
    country_data = {}
    for field_elem in country_elem.findall('field'):
        metric_name = field_elem.get('metric')
        if metric_name in renamed_metrics:
            new_col_name = renamed_metrics[metric_name]
            metric_value = field_elem.text
            if metric_value is None:
                metric_value = np.nan 
            country_data[new_col_name] = metric_value
    if country_data:
            # Ensure the country name is captured for the index later
            country_data['Country Name'] = country_name
            country_datalist.append(country_data)

poverty_data = pd.DataFrame(country_datalist)

if 'Country Name' in poverty_data.columns:
    poverty_data = poverty_data.set_index("Country Name")
    poverty_data.index.name = None
for col in poverty_data.columns:
    poverty_data[col] = pd.to_numeric(poverty_data[col], errors='coerce')

print (poverty_data)




```
