## Usability Analysis

```python
# TODO 1: Place your solution here.

def get_usability(input_file):
    df = pd.read_excel(input_file)
    df['Country Name'] = df.iloc[:, 0]
    df['Missing Values'] = df.isna().sum(axis=1)    
    conditions = [(df['Missing Values'] <= 3), (df['Missing Values'] > 3) & (df['Missing Values'] <= 6), (df['Missing Values'] > 6) & (df['Missing Values'] <= 10), (df['Missing Values'] > 10) & (df['Missing Values'] <= 15), (df['Missing Values'] > 15)]
    values = ['Excellent', 'Good', 'Fair', 'Marginal', 'None']
    df['Usability'] = np.select(conditions, values, default="Unknown")
    output_df = df[['Country Name', 'Missing Values', 'Usability']]
    # print(output_df.head(10).to_string(index=False))
    
    output_df.to_excel('wdi_exports/country-centric-assessment.xlsx', sheet_name='Countries Missing Data', index=False)

get_usability('wdi_data.xlsx')
```

```python
# TODO 2: Place your solution here.
def get_stat_usability(input_file):
    output_df = pd.read_excel(input_file) 
    missing_counts = output_df.isna().sum(axis=0) 
    missing_counts = missing_counts.iloc[1:]
    output_df = pd.DataFrame({
        'Statistic Name': missing_counts.index,
        'Missing Values': missing_counts.values
    }) 
    conditions = [(output_df['Missing Values'] <= 30), (output_df['Missing Values'] > 30) & (output_df['Missing Values'] <= 60), (output_df['Missing Values'] > 60) & (output_df['Missing Values'] <= 100), (output_df['Missing Values'] > 100) & (output_df['Missing Values'] <= 180), (output_df['Missing Values'] > 180)]
    values = ['Excellent', 'Good', 'Fair', 'Marginal', 'None']
    output_df['Usability'] = np.select(conditions, values, default="Unknown")
    output_df = output_df[['Statistic Name', 'Missing Values', 'Usability']]
    print(output_df.head(10).to_string(index=False))
    
    output_df.to_excel('wdi_exports/statistic-centric-assessment.xlsx', sheet_name='Statistics Missing Data', index=False)

get_stat_usability('wdi_data.xlsx')
```

## Descriptive Statistics

```python
# TODO 3: Place your solution here.
def get_descriptive(input_file):
    df = pd.read_excel(input_file)
    country_col = df.columns[0]
    starting_cols = df.columns[1:]
    print(country_col)
    print(starting_cols)
    writer = pd.ExcelWriter("wdi_exports/wdi_summary.xlsx", engine='xlsxwriter')
    for col in starting_cols:
        col_data = df[[country_col, col]].dropna()
        top_10 = col_data.sort_values(by=[col, country_col], ascending=[False, True]).head(10)
        bottom_10 = col_data.sort_values(by=[col, country_col], ascending=[True, True]).head(10)
        max_val = col_data[col].max()
        mean_val = col_data[col].mean()
        min_val = col_data[col].min()
        sheet_rows = [['Max', max_val], ['Mean', mean_val], ['Min', min_val], []]
       
        for i in range(len(top_10)):
            row_name = 'Top 10' if i == 0 else ''
            sheet_rows.append([row_name, top_10.iloc[i][country_col], top_10.iloc[i][col]])
        for i in range(len(bottom_10)):
            row_name = 'Bottom 10' if i == 0 else ''
            sheet_rows.append([row_name, bottom_10.iloc[i][country_col], bottom_10.iloc[i][col]])
        sheet_df = pd.DataFrame(sheet_rows)
        sheet_name = col[:31]
        sheet_df.to_excel(writer, sheet_name=sheet_name, index=False, header=False)

    writer.close()
    
get_descriptive('wdi_data.xlsx')
```
